# Automatic Datatypes

```python
import numpy as np
a = np.array([1,2,3.5,4,5.5])
print(a)
print(a.dtype)
```

# Zeros

```python
import numpy as np
a = np.zeros([5])
print(a)
```

# Integer zeros

```python
import numpy as np
arr = np.zeros(5, dtype=int)
print(arr)
```

```python
# 2d zero
import numpy as np
a = np.zeros((3,4))
print(a)
```
