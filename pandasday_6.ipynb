```python
import pandas as pd
df= pd.read_csv('students.csv')
print(df)
```

```python
print(df.isnull())
```

```python
df.loc[df['Age']<0,'Age']=None
df['Age']=df['Age'].fillna(df['Age'].mean())

df['Marks']=df['Marks'].fillna(df['Marks'].mean())

df['Email']=df['Email'].fillna(df['Name'].str.lower()+'@gmail.com')
print(df)
```

```python
df['Gender']=df['Gender'].replace({
    'Male': 'M',
    'Female': 'F',
    'male' : 'M',
    'female' : 'F',
    'Unknown' :'M'
})
print(df)
```

```python
df['City'] =df['City'].str.strip().str.upper()
    
print(df)
```

```python
df['Name']=df['Name'].str.strip().replace({'prabin':'Prabin'})
print(df)
```

```python
df=df.drop_duplicates(subset=['Name']).reset_index(drop=True)
print(df)
```

```python
print("Duplicate rows:",df.duplicated().sum())
```

```python
import pandas as pd
df= pd.read_csv('employee_data_dirty.csv')
print(df)
```

```python
df.info()
```

```python
df['Age']=df['Age'].fillna(df['Age'].mean())

df['Salary']=df['Salary'].fillna(df['Salary'].mean())

print(df)
```

```python
df['Gender']=df['Gender'].str.title()

df['City']=df['City'].str.strip().str.title()

df['Name']=df['Name'].str.strip().str.title()

print(df)
```

```python
print("Duplicate rows:",df.duplicated().sum())
```

```python
df.info()
```

```python
df= df[df['Email'].str.contains('@', na=False)]
df= df.dropna()
print(df)
```

```python
df['Join_Date'] = pd.to_datetime(
    df['Join_Date'],
    errors='coerce',
    format='mixed'
)

df['Join_Date'] = df['Join_Date'].dt.strftime('%d-%m-%Y')
print(df)
```

```python
df.to_csv('employee_cleaned_data.csv', index=True)
```
