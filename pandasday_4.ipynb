# Data Transformation

```python
data ={
    'Name' : ['Alice','Bob','Charlie','David','Eve'],
    'marks' : [80,90,75,85,95]
}
df = pd.DataFrame(data)
print(df)
```

```python
# add percentage column

df['percentage']=df['marks']/100*100
print(df)
```

```python
# increment marks by 5%

df['marks']=df['marks']*1.05
print(df)
```

# Group By

```python
data ={
    'dept' : ['HR','IT','Finance','Marketing','Sales'],
    'salary' : [50000,75000,60000,65000,70000]
}
df=pd.DataFrame(data)
print(df)
```

```python
print(df.describe())
```

```python
print(df.groupby('dept')['salary'].mean())
```

```python
data ={
    'Name':['Alice','Bob','Charlie','David','Eve'],
    'Age' : [25,30,35,40,45],
    'Gender': ['F','M','M','M','F'],
}
df = pd.DataFrame(data)
print(df)
```

```python
df.groupby('Gender')['Age'].mean()
```

```python
df.groupby('Gender')['Age'].sum()
```

```python
df.groupby('Gender')['Age'].count()
```

# Sorting

```python
data ={
    'Name':['Alice','Bob','Charlie','David','Eve'],
    'Age' : [25,30,35,40,45],
    'City':['New York','Los Angles','Chicago','Houston','Phoenix'],
    'marks': [80,90,75,85,95]
}
df=pd.DataFrame(data)
print(df)
```

```python
# sort by ascending order of marks

print(df.sort_values('marks',ascending=True))
```

```python
# sort by ascending order of city name
print(df.sort_values('City',ascending=True))
```
