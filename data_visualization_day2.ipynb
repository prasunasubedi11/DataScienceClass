# Subpolt

```python
import matplotlib.pyplot as plt

months = ['Jan','Feb','Mar','Apr','May']
sales  = [100,120,150,130,180]


plt.subplot(2,2,1)
plt.plot(months,sales)
plt.title('Line Chart')

plt.subplot(2,2,2)
plt.bar(months,sales)
plt.title('Bar Chart')

plt.subplot(2,2,3)
plt.scatter(range(5),sales)
plt.title('Scatter Plot')

plt.subplot(2,2,4)
plt.hist(sales)
plt.title('Histogram')

plt.tight_layout()

plt.show()
```

# Figures and Axes

```python
fig, ax = plt.subplots()

ax.plot(months, sales)
ax.set_title('Monthly Sales')
ax.set_xlabel('Months')
ax.set_ylabel('Sales')

plt.show()
```

```python
fig, axes = plt.subplots(2,2)

axes[0,0].plot(months,sales)
axes[0,0].set_title('Line')

axes[0,1].bar(months,sales)
axes[0,1].set_title('Bar')

axes[1,0].scatter(range(5),sales)
axes[1,0].set_title('Scatter')

axes[1,1].hist(sales)
axes[1,1].set_title('Histogram')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2,2)

axes[0,0].plot(months,sales)
axes[0,0].set_title('Line')

axes[0,1].bar(months,sales)
axes[0,1].set_title('Bar')

axes[1,0].scatter(range(5),sales)
axes[1,0].set_title('Scatter')

axes[1,1].hist(sales)
axes[1,1].set_title('Histogram')

plt.tight_layout()

plt.savefig('my_charts.png', dpi=300, bbox_inches='tight')
plt.show()
```

# Pandas Visualization

```python
import pandas as pd
import matplotlib.pyplot as plt

data ={
    'Month' : ['Jan','Feb','Mar','Apr','May'],
    'Sales' : [100,120,150,130,180]
}

df= pd.DataFrame(data)

df.plot(
    x='Month',
    y='Sales',
    kind='line'
)

plt.show()
```

# Seaborn Scatter Plot

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('tips')
sns.scatterplot(
    data=df,
    x='total_bill',
    y='tip'
)
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x='total_bill',
    y='tip',
    hue='sex'
)
plt.show()
```

# Seaborn Line Plot

```python
sns.lineplot(
    data=df,
    x='size',
    y='tip'
)

plt.show()
```

```python
sns.barplot(
    data=df,
    x='day',
    y='total_bill',
    errorbar=None
)

plt.show()
```

# Count Plot

```python
# A count plot is useful when we want to know how many observations belong to each category.
# How many record are from:
# Thursday?
# Saturday?
# Sunday?

sns.countplot(
    data=df,
    x='day'
)

plt.show()
```

# Box Plot with Seaborn

```python
sns.boxplot(
    data=df,
    x='day',
    y='total_bill'
)

plt.show()
```
