```python
import pandas as pd
df= pd.read_csv('students.csv')
print(df)
```

```python
df.loc[df['Age']<0,'Age']=None
df['Age']=df['Age'].fillna(df['Age'].mean())
df['Age']=df['Age'].astype(int)

df['Marks']=df['Marks'].fillna(df['Marks'].mean())

df['Email']=df['Email'].fillna(df['Name'].str.lower()+'@gmail.com')

df['Name']=df['Name'].str.strip().str.title()

df['City']=df['City'].str.strip().str.title()
print(df)
```

```python
df['Gender']=df['Gender'].replace({
    'Male': 'M',
    'Female': 'F',
    'male' : 'M',
    'female' : 'F',
    'Unknown' :'M'
})
print(df)
```

```python
print("Duplicate rows:",df.duplicated().sum())
# since id is different for all there is no  any duplicates
```

```python
df= df[df['Email'].str.contains('@', na=False)]
df= df.dropna()
print(df)
```

```python
df= df[pd.to_datetime(df['Join_Date'],errors='coerce').notna()]
print(df)
```

```python
df.to_csv('Cleaned_Student_Data',index=True)
```
