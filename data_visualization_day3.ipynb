# Violin Plot

```python
import matplotlib.pyplot as plt
import seaborn as sns

df=sns.load_dataset('tips')

sns.violinplot(
    data=df,
    x='day',
    y='total_bill'
)

plt.show()
```

```python
# Top tip (Max Bill)
#                 /\
#                /  \
#               |  | |
#               |  | |
#              /   |  \
#             /   [|]  \
#            |     .    |
#            \    [|]   /
#             \    |   /
#              \   |  /
#               \  | /
#                \  /
#                 \/
#         Bottom tip (Min bill)
```

# Heatmap

```python
# Heatmapp represents values using different intensities.

# One of the most important applications in Data Science is correlation visualization.
```

```python
# First calculate correlation:

correlation =df.corr(numeric_only=True)

print(correlation)
```

```python
sns.heatmap(correlation)

plt.show()
```

```python
sns.heatmap(
    correlation,
    annot=True
)
plt.show()
```

# Pairplot

```python
# One very useful Seaborn visualization.
# It creates multiple plots showing relationships between numerical variables.
# Diagonals shows the distribution.
# Off diagonals shows the correlation.

sns.pairplot(df)

plt.show()
```

```python
sns.pairplot(
    df,
    hue='sex'
)
plt.show()
```

```python
sns.histplot(
    data=df,
    x='total_bill'
)
plt.show()
```

```python
sns.histplot(
    data=df,
    x='total_bill',
    kde=True
)
plt.show()

# KDE= Kernel Density Estimate
# It gives a smooth representation of the distribution.
```

# Adding Titles to Seaborn

```python
sns.scatterplot(
    data=df,
    x='total_bill',
    y='tip'
)

plt.title('Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('tip')

plt.show()
```
