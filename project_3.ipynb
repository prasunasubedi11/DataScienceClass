# Practice Iris Dataset

```python
import seaborn as sns
df= sns.load_dataset('iris')
print(df.head())
df.info()
df.describe()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(
    data=df,
    x='sepal_length',
    y='sepal_width'
)
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x='sepal_length',
    y='sepal_width',
    hue='species' # hue means using color to represent different categories in your data
)
plt.show()
```

```python
sns.lineplot(
    data=df,
    x='petal_length',
    y='petal_width'
)

plt.show()

# The shadow area around the line represents the uncertainty or variation in the estimates relationship.
```

```python
sns.barplot(
    data=df,
    x='species',
    y='sepal_length',
    errorbar=None
)
plt.figure(figsize=(12,8))
plt.show()

# Barplot hepls to compare average sepal_length between different species.
```

```python
sns.countplot(
    data=df,
    x='species'
)

plt.show()
```

```python
sns.boxplot(
    data=df,
    x='species',
    y='sepal_length'
)

plt.show()
```

```python
sns.violinplot(
    data=df,
    x='species',
    y='sepal_length'
)

plt.show()

# It provides more information about where the data values are concentrated.
```

```python
correlation =df.corr(numeric_only=True)

print(correlation)
```

```python
sns.heatmap(correlation)

plt.show()

# It is used to show the correlation between numerical variables.
# The color intensity helps us quickly identify the strength of relationship.
```

```python
sns.heatmap(
    correlation,
    annot=True
)
plt.show()
```

```python
sns.pairplot(df)

plt.show()
```

```python
sns.pairplot(
    df,
    hue='species'
)
plt.show()
```

```python
sns.histplot(
    data=df,
    x='sepal_length'
)
plt.show()
```

```python
sns.histplot(
    data=df,
    x='sepal_length',
    kde=True # Kernel Density Estimate
)
plt.show()
```
