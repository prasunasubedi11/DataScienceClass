# Pandas series

```python
import pandas as pd
age = pd.Series([10,20,30,40])
print(age)
```

```python
import pandas as pd
age = pd.Series([10,20,30,40], index = ['Ram','Prasuna','Prabhat','Sita'])
print(age)
```

# Series from dictionary

```python
marks = {
    "Prasuna" : 89,
    "Prabhat" : 88,
    "Ram" : 78
}

s = pd.Series(marks)
print(s)
print(marks["Prasuna"])
```

```python
marks = {
    "Prasuna" : 89,
    "Prabhat" : 88,
    "Ram" : 78
}

s = pd.Series(marks)
print(s["Prasuna"])
```

# Series Operations

```python
marks = pd.Series([99,98,100,78,89])
print(marks.mean())
print(marks.max())
print(marks.min())
print(marks.sum())
print(marks.count())
print(marks.std())
print(marks.sum()/marks.count())
```

```python
print(marks.describe())
```
