```python
import pandas as pd
df= pd.read_csv('employee_data_dirty.csv')
print(df)
```

```python
df.info()
```

```python
df['Age']=df['Age'].fillna(df['Age'].mean()).astype(int)

df['Salary']=df['Salary'].fillna(df['Salary'].mean()).astype(int)

print(df)
```

```python
df['Name']=df['Name'].str.strip().str.title()

df['City']=df['City'].str.strip().str.title()

df['Gender']=df['Gender'].str.lower().replace({
'female':'Female',
'male' : 'Male'
})

print(df)
```

```python
df =df[df['Email'].str.contains('@', na=False)]
df= df.dropna()
print(df)
```

```python
print("Duplicate rows:",df.duplicated().sum())
```

```python
df['Join_Date'] = pd.to_datetime(df['Join_Date'],format='%d-%m-%Y',errors='coerce')
print(df)
```

```python
df.to_csv('Cleaned_Employee_Data', index=True)
```
