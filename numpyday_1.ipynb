# Numpy array

```python
import numpy as np
numbers=np.array([1,2,3,4,5])
print(numbers)
print(type(numbers))
```

# ndarray

```python
# 0 dimesional array

zero_dimensional_array = np.array(42)
print(zero_dimensional_array)
```

```python
# 1 dimesional array

one_dimensional_array = np.array([1,2,3,4,5])
print(one_dimensional_array)
```

```python
# 2 dimesional array

two_dimensional_array = np.array([[1,2,3],[4,5,6]])
print(two_dimensional_array)
```

```python
# 3 dimesional array
# collection of matrices
three_dimensional_array = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(three_dimensional_array)
```

# N-dim

```python
import numpy as np
a = np.array([10,20,30,40,50])
print(a.ndim)
```

```python
b = np.array([[1,2,3],[4,5,6]])
print(b.ndim)
```

```python
c = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(c.ndim)
```

# Shape

```python
a = np.array([10,20,30,40,50])
print(a.shape)
```

```python
b = np.array([[1,2,3],[4,5,6]])
print(b.shape)
```

```python
c = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(c.shape)
```

# Size

```python
b = np.array([[1,2,3],[4,5,6]])
print(b.size)
```

```python
c = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(c.size)
```

# Dtypes

```python
a = np.array([10,20,30,40,50])
print(a.dtype)
```

```python
b = np.array([1.5,2.5,3.5])
print(b.dtype)
```

# Itemsize

```python
# Memory size of each element in bytes

a = np.array([10,20,30,40,50])
print(a.itemsize)
```

# nbytes

```python
# Total memory used by the array in bytes

a = np.array([10,20,30,40,50])
print(a.nbytes)
```
